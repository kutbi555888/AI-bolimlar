```python
# Parametrlar sonini hisoblash
total_params = sum(p.numel() for p in model.parameters())

print(f"Total parameters: {total_params:,}")
```

```python
# Trainable parametrlarni ko‘rish
trainable = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

print(f"Trainable parameters: {trainable:,}")
```

```python
# Tokenizerni alohida o‘rganish
text = "Hello, how are you?"

tokens = tokenizer.tokenize(text)
ids = tokenizer.encode(text)

print(tokens)
print(ids)
```

```python
for token_id in ids:
    print(
        token_id,
        tokenizer.decode([token_id])
    )
```

```python
# EOS ni  ko‘rish

print("EOS token:", tokenizer.eos_token)
print("EOS token ID:", tokenizer.eos_token_id)
```

```python
# Forward passni ko‘rish
import torch

text = "Artificial intelligence is"

inputs = tokenizer(text, return_tensors="pt")

with torch.no_grad():
    outputs = model(**inputs)

print(outputs.logits.shape)
```

```python
# Keyingi tokenni o‘zimiz topamiz

logits = outputs.logits

last_token_logits = logits[0, -1]

next_token_id = torch.argmax(last_token_logits)

print(next_token_id)
print(tokenizer.decode(next_token_id))
```

```python
# Keyin generate() bilan solishtiramiz
output = model.generate(
    **inputs,
    max_new_tokens=30
)

print(tokenizer.decode(output[0]))
```

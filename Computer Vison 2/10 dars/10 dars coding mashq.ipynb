```python
import os
import yaml

yaml_path = r"skin-burn--1\data.yaml"

# Get absolute path of the dataset folder
abs_path = os.path.abspath("skin-burn--1")

# Update yaml with correct absolute paths
with open(yaml_path, "r") as f:
    data = yaml.safe_load(f)

data["path"] = abs_path
data["train"] = "train/images"
data["val"] = "valid/images"
data["test"] = "test/images"

with open(yaml_path, "w") as f:
    yaml.dump(data, f)

print("Updated yaml:")
print(data)
```

```python
from ultralytics import YOLO

model = YOLO("yolo26s-seg.pt")

model.train(
    data=os.path.abspath(yaml_path),  # use absolute path!
    epochs=5,
    imgsz=640,
    project="skin-burn--1",
    name="exp1"
)
```

```python
from ultralytics import YOLO

# Pretrained modelni yukla
model = YOLO('yolo26n-seg.pt')

# Fine-tuning boshlash
results = model.train(
    data=os.path.abspath(yaml_path),     # absolyut pathdan foydalanish
    epochs=3,
    imgsz=512,
    batch=2,
    device='cpu',          # CPUdan foydalanish uchun
    # device='0'     # GPUdan foydalanish uchun
    project='skin-burn--1',
    name='skin_burn',
    patience=5,
)

print("Trening tugadi!")
print(results.save_dir)

# vaqt jua kop ketdi.data ham koproq edi.shuning uchun toxtatdim
```

```python
from ultralytics import YOLO
from PIL import Image
from IPython.display import display       #  Agar Jupyter Notebook’da display tanilmasa

# O'z modelimizni yuklash
model = YOLO('runs/segment/my_model/weights/best.pt')

# Yangi rasmda sinash
results = model.predict(
    source='test_image.jpg',
    conf=0.25,
    save=True,
)

for r in results:
    display(Image.fromarray(r.plot()[..., ::-1]))

# Validation metrikalar
metrics = model.val()

print(f"mAP50      : {metrics.box.map50:.4f}")
print(f"mAP50-95   : {metrics.box.map:.4f}")
print(f"Mask mAP   : {metrics.seg.map:.4f}")
```
